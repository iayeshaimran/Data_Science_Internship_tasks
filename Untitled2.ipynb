```python
# Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
iris = sns.load_dataset('iris')

# First 5 rows
print(iris.head())

# Shape
print("Shape:", iris.shape)

# Columns
print("Columns:", iris.columns)

# Information,
print(iris.info())
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='sepal_length',
    y='petal_length',
    hue='species',
    data=iris
)

plt.title("Sepal Length vs Petal Length")
plt.show()
plt.show


```

```python
plt.figure(figsize=(8,5))

plt.hist(iris['sepal_length'])

plt.title("Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='species',
    y='sepal_length',
    data=iris
)

plt.title("Box Plot of Sepal Length")
plt.show()
```

The Iris dataset was loaded and analyzed successfully. Scatter plot showed relationships between variables, histogram showed data distribution, and box plot helped identify spread and outliers in the data
